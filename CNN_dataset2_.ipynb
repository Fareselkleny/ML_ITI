```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras.datasets import cifar10
from tensorflow.keras.utils import to_categorical
from tensorflow.keras import layers, models
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten
from tensorflow.keras.layers import Input, Dense, Dropout, BatchNormalization
from tensorflow.keras.models import Model, load_model
from tensorflow.keras.optimizers import RMSprop
from tensorflow.keras.callbacks import ReduceLROnPlateau, EarlyStopping, ModelCheckpoint

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

from sklearn.preprocessing import MinMaxScaler

np.random.seed(42)
tf.random.set_seed(42)

(x_train, y_train), (x_test, y_test) = cifar10.load_data()

x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)

x_train, x_val, y_train, y_val = train_test_split(
    x_train,
    y_train,
    test_size=0.1,
    random_state=42,
    stratify=np.argmax(y_train, axis=1)
)

print("x_train shape:", x_train.shape)
print("x_val shape:", x_val.shape)
print("x_test shape:", x_test.shape)
print("y_train shape:", y_train.shape)
print("y_val shape:", y_val.shape)
print("y_test shape:", y_test.shape)

(_, _), (_, labels) = cifar10.load_data()

idx = [3, 6, 25, 46, 58, 85, 93, 99, 108, 133]

clsmap = {
    0: "airplane",
    1: "automobile",
    2: "bird",
    3: "cat",
    4: "deer",
    5: "dog",
    6: "frog",
    7: "horse",
    8: "ship",
    9: "truck"
}

plt.figure(figsize=(10, 4))

for i, index in enumerate(idx):

    plt.subplot(2, 5, i + 1)

    plt.imshow(x_test[index])

    label = labels[index][0]

    plt.title(
        str(label) + ": " + clsmap[label]
    )

    plt.xticks([])
    plt.yticks([])

plt.tight_layout()
plt.show()

inpt_dim = (32, 32, 3)

inpt_img = Input(
    shape=inpt_dim,
    name="input"
)

c11 = Conv2D(
    64,
    (9, 9),
    strides=(2, 2),
    activation="relu",
    name="conv1"
)(inpt_img)

p12 = MaxPooling2D(
    pool_size=(2, 2),
    name="pool1"
)(c11)

bn13 = BatchNormalization(
    name="batchnorm1"
)(p12)

c14 = Conv2D(
    128,
    (3, 3),
    strides=(1, 1),
    activation="relu",
    name="conv2"
)(bn13)

p15 = MaxPooling2D(
    pool_size=(2, 2),
    name="pool2"
)(c14)

bn16 = BatchNormalization(
    name="batchnorm2"
)(p15)

f17 = Flatten(
    name="flatten"
)(bn16)

do18 = Dropout(
    0.5,
    name="dropout1"
)(f17)

d19 = Dense(
    256,
    activation="relu",
    name="dense1"
)(do18)

do110 = Dropout(
    0.2,
    name="dropout2"
)(d19)

d111 = Dense(
    64,
    activation="relu",
    name="dense2"
)(do110)

do112 = Dropout(
    0.1,
    name="dropout3"
)(d111)

output = Dense(
    10,
    activation="softmax",
    name="output"
)(do112)

classifier = Model(
    inputs=inpt_img,
    outputs=output
)

opt = RMSprop(
    learning_rate=0.001
)

classifier.compile(
    optimizer=opt,
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

print(classifier.summary())

reduce_lr = ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.5,
    patience=5,
    min_delta=1e-4,
    mode="min",
    verbose=1,
    min_lr=1e-6
)

stop_alg = EarlyStopping(
    monitor="val_loss",
    patience=10,
    restore_best_weights=True,
    verbose=1
)

checkpoint_filepath = "cnn_model.keras"

model_checkpoint = ModelCheckpoint(
    filepath=checkpoint_filepath,
    monitor="val_accuracy",
    mode="max",
    save_best_only=True,
    verbose=1
)

hist = classifier.fit(
    x_train,
    y_train,
    batch_size=128,
    epochs=30,
    callbacks=[
        stop_alg,
        reduce_lr,
        model_checkpoint
    ],
    shuffle=True,
    validation_data=(
        x_val,
        y_val
    )
)

classifier.save_weights(
    "cnn1.weights.h5"
)

best_model = load_model(
    "cnn_model.keras"
)

print("\nBest model loaded successfully.")

plt.figure(figsize=(10, 6))

plt.plot(
    hist.history["loss"],
    label="Training Loss"
)

plt.plot(
    hist.history["val_loss"],
    label="Validation Loss"
)

plt.title(
    "Model Loss Progress"
)

plt.ylabel(
    "Categorical Cross Entropy Loss"
)

plt.xlabel(
    "Epoch"
)

plt.legend(
    ["Training Set", "Validation Set"],
    loc="upper right"
)

plt.grid()

plt.show()

plt.figure(figsize=(10, 6))

plt.plot(
    hist.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    hist.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.title(
    "Model Accuracy Progress"
)

plt.ylabel(
    "Accuracy"
)

plt.xlabel(
    "Epoch"
)

plt.legend(
    ["Training Set", "Validation Set"],
    loc="lower right"
)

plt.grid()

plt.show()

test_loss, test_accuracy = best_model.evaluate(
    x_test,
    y_test,
    batch_size=128,
    verbose=1
)

print("\n==============================")
print("       TEST RESULTS")
print("==============================")

print(
    "Test Loss:",
    test_loss
)

print(
    "Test Accuracy:",
    test_accuracy
)

print(
    "Test Accuracy (%):",
    test_accuracy * 100
)

y_true = np.argmax(
    y_test,
    axis=1
)

y_hat = best_model.predict(
    x_test,
    batch_size=128,
    verbose=1
)

y_pred = np.argmax(
    y_hat,
    axis=1
)

accuracy = np.mean(
    y_true == y_pred
)

precision = precision_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

recall = recall_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

f1 = f1_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

balanced_accuracy = balanced_accuracy_score(
    y_true,
    y_pred
)

ber = 1 - balanced_accuracy

print("\n==============================")
print("    EVALUATION METRICS")
print("==============================")

print(
    "Accuracy           :",
    accuracy
)

print(
    "Accuracy (%)       :",
    accuracy * 100
)

print(
    "Precision          :",
    precision
)

print(
    "Recall             :",
    recall
)

print(
    "F1-Score           :",
    f1
)

print(
    "Balanced Accuracy  :",
    balanced_accuracy
)

print(
    "BER                :",
    ber
)

print("\n==============================")
print("    CLASSIFICATION REPORT")
print("==============================")

print(
    classification_report(
        y_true,
        y_pred,
        labels=list(range(10)),
        target_names=[
            clsmap[i]
            for i in range(10)
        ],
        zero_division=0
    )
)

cm = confusion_matrix(
    y_true,
    y_pred,
    labels=list(range(10))
)

print("\nConfusion Matrix:")
print(cm)

plt.figure(
    figsize=(10, 8)
)

plt.imshow(
    cm,
    interpolation="nearest"
)

plt.title(
    "Confusion Matrix"
)

plt.colorbar()

tick_marks = np.arange(10)

plt.xticks(
    tick_marks,
    [clsmap[i] for i in range(10)],
    rotation=45
)

plt.yticks(
    tick_marks,
    [clsmap[i] for i in range(10)]
)

threshold = cm.max() / 2.0

for i in range(cm.shape[0]):

    for j in range(cm.shape[1]):

        plt.text(
            j,
            i,
            format(cm[i, j], "d"),
            horizontalalignment="center",
            color="white" if cm[i, j] > threshold else "black"
        )

plt.ylabel(
    "True Label"
)

plt.xlabel(
    "Predicted Label"
)

plt.tight_layout()

plt.show()

idx = [
    3,
    6,
    25,
    46,
    58,
    85,
    93,
    99,
    108,
    133
]

predictions = best_model.predict(
    x_test[idx],
    verbose=0
)

predicted_labels = np.argmax(
    predictions,
    axis=1
)

plt.figure(
    figsize=(12, 6)
)

for i in range(10):

    plt.subplot(
        2,
        5,
        i + 1
    )

    plt.imshow(
        x_test[idx[i]]
    )

    true_class = np.argmax(
        y_test[idx[i]]
    )

    predicted_class = predicted_labels[i]

    confidence = (
        predictions[i][predicted_class] * 100
    )

    plt.title(
        f"True: {clsmap[true_class]}\n"
        f"Pred: {clsmap[predicted_class]}\n"
        f"{confidence:.1f}%"
    )

    plt.xticks([])
    plt.yticks([])

plt.tight_layout()
plt.show()

first_conv = best_model.get_layer(
    "conv1"
)

w = first_conv.get_weights()[0]

wshape = w.shape

scaler = MinMaxScaler()

scaler.fit(
    w.reshape(-1, 1)
)

w_scaled = scaler.transform(
    w.reshape(-1, 1)
)

w_scaled = w_scaled.reshape(
    wshape
)

num_filters = w_scaled.shape[-1]

fig, axs = plt.subplots(
    8,
    8,
    figsize=(18, 18)
)

axs = axs.ravel()

for i in range(64):

    h = w_scaled[:, :, :, i]

    axs[i].imshow(
        h
    )

    axs[i].set_title(
        "Filter " + str(i)
    )

    axs[i].axis("off")

plt.tight_layout()
plt.show()

print("\n==============================")
print("      SAMPLE PREDICTIONS")
print("==============================")

for i in range(10):

    true_class = np.argmax(
        y_test[idx[i]]
    )

    predicted_class = predicted_labels[i]

    confidence = (
        predictions[i][predicted_class] * 100
    )

    print(
        f"Image {idx[i]} | "
        f"True: {clsmap[true_class]} | "
        f"Predicted: {clsmap[predicted_class]} | "
        f"Confidence: {confidence:.2f}%"
    )

print("\n==============================")
print("      MODEL INFORMATION")
print("==============================")

print(
    "Model saved as: cnn_model.keras"
)

print(
    "Weights saved as: cnn1.weights.h5"
)

print(
    "Number of classes: 10"
)

print(
    "Input shape: 32x32x3"
)

print(
    "Optimizer: RMSprop"
)

print(
    "Loss: Categorical Crossentropy"
)

print(
    "Activation: Softmax"
)
```
